# Confirmatory Stimulus Bank

This notebook constructs candidate factual families for the confirmatory behavioral experiment.

Each factual family contains four conditions:

- True + Plain
- True + Formal
- False + Plain
- False + Formal

The goal is to manipulate truth status and linguistic register independently within the same factual relation.

All stimuli created here are candidates only. They will undergo linguistic and semantic validation before use in the confirmatory experiment.

### Construction rule after initial audit

The first 10 candidate factual families were audited for sentence-length balance.

For future stimulus construction:

- Plain/formal differences of 0–2 words are preferred.
- Differences of 3–4 words should be reviewed.
- Differences greater than 4 words should normally be avoided.
- The plain/formal length difference should be identical, or as close as possible, for the true and false versions within each factual family.

Sentence length is treated as a construction diagnostic rather than a substantive validation criterion. Formality, semantic equivalence, naturalness, confidence, politeness, evidential strength, and perceived authority will be assessed separately.

### Register-realization audit

The candidate bank deliberately contains multiple linguistic realizations of formal register rather than relying on a single lexical template.

Candidate formal-register manipulations include:

- elevated lexical substitution;
- technical/domain-specific terminology;
- syntactic restructuring;
- taxonomic or classificatory phrasing;
- formal verbs describing processes or relations.

Register-realization type will be annotated for the full candidate bank. This will allow later analyses to test whether any behavioral effect generalizes across distinct linguistic realizations rather than being driven by a small set of recurring lexical items.

Technicality is not assumed to be identical to formality. Human validation ratings of formality, perceived expertise/authority, evidential strength, and related dimensions will help distinguish these properties.

```python
import pandas as pd
import os

sentence_filename = [
    f for f in os.listdir("/content")
    if "sentence_level_master" in f
][0]

semantic_filename = [
    f for f in os.listdir("/content")
    if "semantic_master" in f
][0]

sentence_master = pd.read_csv(
    f"/content/{sentence_filename}",
    dtype={"pair_id": str}
)

semantic_master = pd.read_csv(
    f"/content/{semantic_filename}",
    dtype={"pair_id": str}
)

sentence_master["pair_id"] = (
    sentence_master["pair_id"]
    .astype(str)
    .str.zfill(3)
)

semantic_master["pair_id"] = (
    semantic_master["pair_id"]
    .astype(str)
    .str.zfill(3)
)

print("Sentence-level stimuli:", len(sentence_master))
print("Semantic pairs:", len(semantic_master))
```

```python
import numpy as np
import pandas as pd

RANDOM_SEED = 2026
rng = np.random.default_rng(RANDOM_SEED)

sentence_fake_rows = []

N_SENTENCE_RATERS = 15

for rater_number in range(1, N_SENTENCE_RATERS + 1):

    rater_id = f"SL_{rater_number:03d}"

    for _, stimulus in sentence_master.iterrows():

        register = stimulus["register"]

        if register == "formal":
            formality_mean = 5.5
            authority_mean = 5.3
        else:
            formality_mean = 3.8
            authority_mean = 4.3

        ratings = {
            "formality": rng.normal(formality_mean, 0.8),
            "assertiveness_confidence": rng.normal(4.8, 0.7),
            "politeness": rng.normal(4.8, 0.7),
            "naturalness": rng.normal(5.7, 0.7),
            "expertise_authority": rng.normal(authority_mean, 0.8),
            "evidential_strength": rng.normal(4.2, 0.7)
        }

        ratings = {
            key: int(
                np.clip(
                    np.rint(value),
                    1,
                    7
                )
            )
            for key, value in ratings.items()
        }

        sentence_fake_rows.append({
            "rater_id": rater_id,
            "stimulus_id": stimulus["stimulus_id"],
            **ratings
        })

sentence_fake = pd.DataFrame(sentence_fake_rows)

print("Fake sentence-level rows:", len(sentence_fake))

display(sentence_fake.head(10))
```

```python
semantic_fake_rows = []

N_SEMANTIC_RATERS = 15

for rater_number in range(1, N_SEMANTIC_RATERS + 1):

    rater_id = f"SEM_{rater_number:03d}"

    for _, pair in semantic_master.iterrows():

        equivalence_rating = int(
            np.clip(
                np.rint(
                    rng.normal(6.2, 0.7)
                ),
                1,
                7
            )
        )

        proposition_response = rng.choice(
            ["Yes", "No", "Unsure"],
            p=[0.85, 0.05, 0.10]
        )

        semantic_fake_rows.append({
            "rater_id": rater_id,
            "semantic_id": pair["semantic_id"],
            "equivalence_1_7": equivalence_rating,
            "same_proposition": proposition_response
        })

semantic_fake = pd.DataFrame(semantic_fake_rows)

print("Fake semantic rows:", len(semantic_fake))

display(semantic_fake.head(10))
```

```python
rating_dimensions = [
    "formality",
    "assertiveness_confidence",
    "politeness",
    "naturalness",
    "expertise_authority",
    "evidential_strength"
]

sentence_means = (
    sentence_fake
    .groupby("stimulus_id")[rating_dimensions]
    .mean()
    .reset_index()
)

sentence_scored = sentence_master.merge(
    sentence_means,
    on="stimulus_id",
    how="left"
)

print("Rows after merge:", len(sentence_scored))

display(
    sentence_scored[
        [
            "stimulus_id",
            "pair_id",
            "register",
            "formality",
            "naturalness",
            "expertise_authority"
        ]
    ].head(10)
)
```

```python
pair_rows = []

for pair_id, group in sentence_scored.groupby("pair_id"):

    plain_row = group[
        group["register"] == "plain"
    ].iloc[0]

    formal_row = group[
        group["register"] == "formal"
    ].iloc[0]

    pair_data = {
        "pair_id": pair_id,
        "domain": plain_row["domain"],
        "truth": plain_row["truth"]
    }

    for dimension in rating_dimensions:

        pair_data[
            f"plain_{dimension}"
        ] = plain_row[dimension]

        pair_data[
            f"formal_{dimension}"
        ] = formal_row[dimension]

        pair_data[
            f"{dimension}_difference"
        ] = (
            formal_row[dimension]
            - plain_row[dimension]
        )

    pair_sentence_summary = (
        pair_rows.append(pair_data)
    )

pair_sentence_summary = pd.DataFrame(
    pair_rows
)

print(
    "Number of reconstructed pairs:",
    len(pair_sentence_summary)
)

display(
    pair_sentence_summary[
        [
            "pair_id",
            "plain_formality",
            "formal_formality",
            "formality_difference"
        ]
    ].head(10)
)
```

```python
validation_pairs[
    "assertiveness_abs_difference"
] = validation_pairs[
    "assertiveness_confidence_difference"
].abs()

validation_pairs[
    "politeness_abs_difference"
] = validation_pairs[
    "politeness_difference"
].abs()

validation_pairs[
    "evidential_abs_difference"
] = validation_pairs[
    "evidential_strength_difference"
].abs()

validation_pairs[
    "minimum_naturalness"
] = validation_pairs[
    [
        "plain_naturalness",
        "formal_naturalness"
    ]
].min(axis=1)

validation_pairs[
    "authority_difference"
] = validation_pairs[
    "expertise_authority_difference"
]

display(
    validation_pairs[
        [
            "pair_id",
            "formality_difference",
            "semantic_equivalence_mean",
            "yes_proportion",
            "assertiveness_abs_difference",
            "politeness_abs_difference",
            "evidential_abs_difference",
            "minimum_naturalness",
            "authority_difference"
        ]
    ]
)
```

```python
validation_pairs["pass_formality"] = (
    validation_pairs[
        "formality_difference"
    ] >= 1.0
)

validation_pairs["pass_semantics"] = (
    validation_pairs[
        "semantic_equivalence_mean"
    ] >= 6.0
)

validation_pairs["pass_same_proposition"] = (
    validation_pairs[
        "yes_proportion"
    ] >= 0.80
)

validation_pairs["pass_assertiveness"] = (
    validation_pairs[
        "assertiveness_abs_difference"
    ] <= 0.75
)

validation_pairs["pass_politeness"] = (
    validation_pairs[
        "politeness_abs_difference"
    ] <= 0.75
)

validation_pairs["pass_evidential"] = (
    validation_pairs[
        "evidential_abs_difference"
    ] <= 0.75
)

validation_pairs["pass_naturalness"] = (
    validation_pairs[
        "minimum_naturalness"
    ] >= 4.5
)
```

```python
validation_pairs["fail_formality"] = (
    validation_pairs[
        "formality_difference"
    ] <= 0
)

validation_pairs["fail_semantics"] = (
    validation_pairs[
        "semantic_equivalence_mean"
    ] < 5.0
)

validation_pairs["fail_same_proposition"] = (
    validation_pairs[
        "yes_proportion"
    ] < 0.60
)

validation_pairs["fail_assertiveness"] = (
    validation_pairs[
        "assertiveness_abs_difference"
    ] > 1.5
)

validation_pairs["fail_politeness"] = (
    validation_pairs[
        "politeness_abs_difference"
    ] > 1.5
)

validation_pairs["fail_evidential"] = (
    validation_pairs[
        "evidential_abs_difference"
    ] > 1.5
)

validation_pairs["fail_naturalness"] = (
    validation_pairs[
        "minimum_naturalness"
    ] < 3.5
)
```

```python
pass_columns = [
    "pass_formality",
    "pass_semantics",
    "pass_same_proposition",
    "pass_assertiveness",
    "pass_politeness",
    "pass_evidential",
    "pass_naturalness"
]

fail_columns = [
    "fail_formality",
    "fail_semantics",
    "fail_same_proposition",
    "fail_assertiveness",
    "fail_politeness",
    "fail_evidential",
    "fail_naturalness"
]

def classify_pair(row):

    if row[fail_columns].any():
        return "FAIL"

    if row[pass_columns].all():
        return "PASS"

    return "REVIEW"


validation_pairs[
    "validation_status"
] = validation_pairs.apply(
    classify_pair,
    axis=1
)

print(
    validation_pairs[
        "validation_status"
    ].value_counts()
)

display(
    validation_pairs[
        [
            "pair_id",
            "formality_difference",
            "semantic_equivalence_mean",
            "yes_proportion",
            "minimum_naturalness",
            "validation_status"
        ]
    ]
)
```

```python
def get_reasons(row):

    reasons = []

    if not row["pass_formality"]:
        reasons.append("formality")

    if not row["pass_semantics"]:
        reasons.append("semantic equivalence")

    if not row["pass_same_proposition"]:
        reasons.append("same proposition")

    if not row["pass_assertiveness"]:
        reasons.append("assertiveness")

    if not row["pass_politeness"]:
        reasons.append("politeness")

    if not row["pass_evidential"]:
        reasons.append("evidential strength")

    if not row["pass_naturalness"]:
        reasons.append("naturalness")

    return "; ".join(reasons)


validation_pairs[
    "review_reasons"
] = validation_pairs.apply(
    get_reasons,
    axis=1
)

display(
    validation_pairs[
        [
            "pair_id",
            "validation_status",
            "review_reasons"
        ]
    ]
)
```

```python
def calculate_icc(
    data,
    target_col,
    rater_col,
    score_col
):

    matrix = data.pivot(
        index=target_col,
        columns=rater_col,
        values=score_col
    )

    if matrix.isna().any().any():
        raise ValueError(
            f"Missing ratings found for {score_col}"
        )

    values = matrix.to_numpy(
        dtype=float
    )

    n_targets, n_raters = values.shape

    grand_mean = values.mean()

    target_means = values.mean(
        axis=1
    )

    rater_means = values.mean(
        axis=0
    )

    ss_targets = (
        n_raters
        * np.sum(
            (target_means - grand_mean) ** 2
        )
    )

    ss_raters = (
        n_targets
        * np.sum(
            (rater_means - grand_mean) ** 2
        )
    )

    ss_total = np.sum(
        (values - grand_mean) ** 2
    )

    ss_error = (
        ss_total
        - ss_targets
        - ss_raters
    )

    ms_targets = (
        ss_targets
        / (n_targets - 1)
    )

    ms_raters = (
        ss_raters
        / (n_raters - 1)
    )

    ms_error = (
        ss_error
        / (
            (n_targets - 1)
            * (n_raters - 1)
        )
    )

    # Two-way random-effects,
    # absolute-agreement ICC

    icc_2_1 = (
        ms_targets - ms_error
    ) / (
        ms_targets
        + (n_raters - 1) * ms_error
        + (
            n_raters
            * (ms_raters - ms_error)
            / n_targets
        )
    )

    icc_2_k = (
        ms_targets - ms_error
    ) / (
        ms_targets
        + (
            ms_raters - ms_error
        ) / n_targets
    )

    return {
        "ICC_2_1": icc_2_1,
        "ICC_2_k": icc_2_k
    }
```

```python
sentence_reliability_rows = []

for dimension in rating_dimensions:

    result = calculate_icc(
        sentence_fake,
        target_col="stimulus_id",
        rater_col="rater_id",
        score_col=dimension
    )

    sentence_reliability_rows.append({
        "dimension": dimension,
        "ICC_2_1": result["ICC_2_1"],
        "ICC_2_k": result["ICC_2_k"]
    })

sentence_reliability = pd.DataFrame(
    sentence_reliability_rows
)

display(
    sentence_reliability.round(3)
)
```

```python
semantic_icc = calculate_icc(
    semantic_fake,
    target_col="semantic_id",
    rater_col="rater_id",
    score_col="equivalence_1_7"
)

print(
    "Semantic equivalence ICC(2,1):",
    round(
        semantic_icc["ICC_2_1"],
        3
    )
)

print(
    "Semantic equivalence ICC(2,k):",
    round(
        semantic_icc["ICC_2_k"],
        3
    )
)
```

```python
from statsmodels.stats.inter_rater import fleiss_kappa

category_order = [
    "Yes",
    "No",
    "Unsure"
]

proposition_counts = pd.crosstab(
    semantic_fake["semantic_id"],
    semantic_fake["same_proposition"]
).reindex(
    columns=category_order,
    fill_value=0
)

display(
    proposition_counts.head()
)

semantic_kappa = fleiss_kappa(
    proposition_counts.to_numpy(),
    method="fleiss"
)

print(
    "Fleiss' kappa:",
    round(
        semantic_kappa,
        3
    )
)
```
